# 福井大学 説明資料

## 1. 概要
- 福井大学（University of Fukui）は、福井県福井市に本部を置く国立大学です。
- 1949年に設立され、地域社会と連携しながら教育・研究を行っています。

## 2. 学部・学科
- 教育学部
- 医学部
- 工学部
- 国際地域学部

## 3. 特徴
- 地域貢献型の教育・研究
- 医学・工学・教育の連携
- 国際交流プログラムの充実

## 4. キャンパス
- 文京キャンパス（教育学部・工学部・国際地域学部）
- 松岡キャンパス（医学部）

## 5. 主な研究分野
- 医療・健康科学
- 環境・エネルギー
- 教育・人間科学
- 地域創生

## 6. 国際交流
- 多くの海外大学と提携
- 留学生受け入れ・派遣プログラム

## 7. アクセス
- JR福井駅からバスで約15分（文京キャンパス）

## 8. 公式サイト
- https://www.u-fukui.ac.jp/
